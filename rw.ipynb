```python
my_string = "Hello World!"

# Convert to upper case
print(my_string.upper()) # Output: HELLO WORLD!

# Convert to lower case
print(my_string.lower()) # Output: hello world!

# Remove whitespace
print(my_string.strip()) # Output: Hello World!

# Replacing String
print(my_string.replace("World", "Universe")) # Output: Hello Universe!

# Splitting String
print(my_string.split(" ")) # Output: ['Hello', ' World!']
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(6))
```

```python
def countdown(n):
    if n == 0:
        print("Done")
    else:
        print(n)
        countdown(n - 1)

countdown(10)
```

```python
print("Python Data Analytics Lab Test")
import csv
try:
    with open("orders.csv", "r") as file:
        reader = csv.reader(file)
        header = next(reader)
        records = []
        for row in reader:
            records.append(row)
    print("Header:")
    print(header)
    print("Records:")
    for row in records:
        print(row)
    processed_records = []
    for row in records:
        price = float(row[5])
        quantity = int(row[6])
        total_amount = price * quantity
        new_row = row.copy()
        new_row[5] = price
        new_row[6] = quantity
        new_row.append(total_amount)
        processed_records.append(new_row)
    new_header = header.copy()
    new_header.append("total_amount")
    print("Updated Header:")
    print(new_header)
    print("Processed Records:")
    for row in processed_records:
        print(row)
except FileNotFoundError:
    print("Error: orders.csv file not found")
```
