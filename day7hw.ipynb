```python

import csv
try:
    with open("sales.csv", "r", newline="") as file:
        reader = csv.reader(file)
        next(reader)
        summary = []
        for row in reader:
            month = row[0]
            amount_text = row[1]
            try:
                amount = int(amount_text)
                if amount < 0:
                    raise ValueError("Negative sales amount")
            except ValueError as error:
                print(f"Skipping {month}: {error}")
                continue
            if amount >= 80000:
                category = "High"
            elif amount >= 50000:
                category = "Medium"
            else:
                category = "Low"
            summary.append([month, amount, category])
    with open("sales_summary.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["month", "amount", "category"])
        writer.writerows(summary)
    print("Sales summary created successfully!")
except FileNotFoundError:
    print("Error: sales.csv file not found.")
```
