```python
valid_marks = []
try:
    with open("marks.txt", "r") as file:
        for line in file:
            marks = line.split()
            for value in marks:
                try:
                    mark = int(value)
                except ValueError:
                    print("Invalid data skipped:", value)
                    continue
                try:
                    if mark < 0 or mark > 100:
                        raise ValueError("Mark out of range")
                except ValueError as error:
                    print("Invalid mark skipped:", value)
                    continue
                else:
                    valid_marks.append(mark)
    valid_count = 0
    pass_count = 0
    fail_count = 0
    for mark in valid_marks:
        valid_count += 1
        if mark >= 40:
            pass_count += 1
        else:
            fail_count += 1
except FileNotFoundError:
    print("File not found")
else:
    print("Successfully processed!")
    print("Valid marks:", valid_count)
    print("Pass marks:", pass_count)
    print("Fail marks:", fail_count)
finally:
    print("Analysis process completed")
```
