```python
students = int(input("Total students:"))
total = 0
passed = 0
failed = 0
for i in range(students):
    marks = int(input("marks:"))
    if marks == -1:
        break
    if marks < 0 or marks > 100:
        print("invalid mark, skipped")
        continue
    total += 1
    print("valid")
    if marks >= 80:
        print("Excellent")
        passed += 1
    elif marks >= 60:
        print("Good")
        passed += 1
    elif marks >= 40:
        print("Pass")
        passed += 1
    else:
        print("Fail")
        failed += 1
print("total", total)
print("passed", passed) 
print("failed", failed)



```
