```python
months = int(input("Number of months: "))
sales = int(input("Sales per month: "))
total = months * sales
for value in sales:
    if value == 0:
        continue
    elif value < 0:
        break
    

def classify_sales(amount):
    if amount >= 100000:
        return "High Performance"
    elif amount >= 50000:
        return "Average Performance"
    else:
        return "Low Performance"
calculate_bonus = lambda amount: amount * 0.10
print("Total Sales:", total)
print("Performance:", classify_sales(total))
print("Bonus:", calculate_bonus(total))

```
