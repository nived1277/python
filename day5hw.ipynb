```python
mobile_buyers = {"Arun", "Priya", "Kumar", "Divya", "Rahul"}
laptop_buyers = {"Priya", "Kumar", "Sneha", "Vijay", "Arun"}
print("Bought both mobile and laptop:", (mobile_buyers).intersection(laptop_buyers))
print("Bought mobile only:", mobile_buyers)
print("mobile buyers:", mobile_buyers - laptop_buyers)
print("laptop buyers:", laptop_buyers - mobile_buyers)
total_customers = mobile_buyers.union(laptop_buyers)
print(total_customers)
print(total_customers.remove("Arun"))
print(total_customers)
```
