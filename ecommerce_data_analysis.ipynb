```python
import csv
try:
    with open("orders.csv", "r") as file:
        reader = csv.reader(file)
        header = next(reader)
        records = []
        for row in reader:
            records.append(row)
    print("Header:")
    print(header)
    print("\nRecords:")
    for row in records:
        print(row)
except FileNotFoundError:
    print("Error: orders.csv file not found")
```

```python
processed_records = []
for row in records:
    price = float(row[5])
    quantity = int(row[6])
    total_amount = price * quantity
    new_row = row.copy()
    new_row[5] = price
    new_row[6] = quantity
    new_row.append(total_amount)
    processed_records.append(new_row)
new_header = header.copy()
new_header.append("total_amount")
print("Updated Header:")
print(new_header)
print("\nProcessed Records:")
for row in processed_records:
    print(row)
```

```python
missing_customers = []
for row in processed_records:
    for i in range(len(row)):
        if isinstance(row[i], str):
            row[i] = row[i].strip()
    if row[1] == "":
        missing_customers.append(row[0])
        print("Missing name:", row[0])
    if row[7] == "":
        row[7] = "Pending"
    row[2] = row[2].lower()
print("Missing customer orders:", missing_customers)
print("Cleaned Records:")
for row in processed_records:
    print(row)
```

```python
print("Product Categories:")
for row in processed_records:
    print(row[4])
print("Products Contains word TOP:")
for row in processed_records:
    if "top" in row[3]:
        print(row[3])
print("Customer name to Upper Case:")
for row in processed_records:
    row[1] = row[1].upper()
    print(row[1])
print("Years from Order Dates:")
for row in processed_records:
    year = row[8].split("-")[0]
    print(year)
```

```python
print("Price exceeds 30000")
for row in processed_records:
    if row[5] > 30000:
        print(row)
print("With Cancelled status")
for row in processed_records:
    if row[7] == "Cancelled":
        print(row)
print("Categorize Order")
for row in processed_records:
    total_amount = row[9]
    if total_amount >= 30000:
        category = "High Value"
    elif total_amount >= 10000:
        category = "Medium Value"
    else:
        category = "Low Value"
    print(row[0], category, total_amount)
```

```python
print("Extract product names")
products = []
for row in processed_records:
    products.append(row[3])
print(products)
print("Identify unique cities")
cities = []
for row in processed_records:
    if row[2] not in cities:
        cities.append(row[2])
print(cities)
print("\nCount total orders:--", len(processed_records))
print("\nSort orders based on price")
sorted_order = sorted(processed_records, key = lambda row:row[5])
for row in sorted_order:
    print(row)
```

```python
city_orders = {}
for row in processed_records:
    city = row[2]
    if city in city_orders:
        city_orders[city] += 1
    else:
        city_orders[city] = 1
print("Order Count per City:")
print(city_orders)
category_orders = {}
for row in processed_records:
    category = row[4]
    if category in category_orders:
        category_orders[category] += 1
    else:
        category_orders[category] = 1
print("\nOrder Count per Category:")
print(category_orders)
category_revenue = {}
for row in processed_records:
    category = row[4]
    total_amount = row[9]
    if category in category_revenue:
        category_revenue[category] += total_amount
    else:
        category_revenue[category] = total_amount
print("\nTotal Revenue per Category:")
print(category_revenue)

```

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total
print(calculate_total(2500,3))
def get_high_value_orders(orders, threshold):
    high_value_orders = []
    for row in orders:
        if row[9] > threshold:
            high_value_orders.append(row)
    return high_value_orders
high_orders = get_high_value_orders(processed_records, 30000)
print("\nHigh Value Orders:")
for row in high_orders:
    print(row)
def orders_by_city(orders, city):
    city_orders = []
    for row in orders:
        if row[2] == city:
            city_orders.append(row)
    return city_orders
delhi_orders = orders_by_city(processed_records, "delhi")
print("\nOrders from Delhi:")
for row in delhi_orders:
    print(row)
```

```python
import csv
try:
    with open("orders.csv", "r") as file:
        reader = csv.reader(file)
        header = next(reader)
        records = []
        for row in reader:
            records.append(row)
    print("File read successfully!")
    for row in records:
        try:
            if row[5] == "" or row[6] == "":
                raise ValueError("Missing price or quantity")
            price = float(row[5])
            quantity = int(row[6])
            total = price * quantity
            print("Order ID:", row[0], "| Total:", total)
        except ValueError as e:
            print("Data Error in Order:", row[0], "-", e)
    else:
        print("\nAll records have been checked.")
except FileNotFoundError:
    print("Error: orders.csv file not found.")
except PermissionError:
    print("Error: Permission denied while accessing the file.")
finally:
    print("\nError handling process completed.")
```
