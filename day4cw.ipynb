```python
marks = [78, 85, 62, 90, 55]
print(marks)
print(len(marks))
print(marks[0])
print(marks[4])
print(marks[1:3])
marks[2] = 70
print(marks)
marks[1:3] = [88,92]
print(marks)
marks.append(76)
print(marks)
marks.insert(1,60)
print(marks)
marks.extend([81,84])
print(marks)
marks.remove(55)
marks.pop(7)
print(marks)
```

```python
for mark in marks:
    print(mark)
```

```python
passed_mark = [mark for mark in marks if mark >= 60]
print("passed mark :", passed_mark)
```

```python
passed_mark.sort()
print(passed_mark)
passed_mark.sort(reverse = True)
print(passed_mark)
```
